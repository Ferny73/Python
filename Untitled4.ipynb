Imagina que un grupo de amigos quiere saber más sobre su cosecha de zanahorias. Cosecha 10 zanahorias y quiere saber y analizar su peso para ver como es su producción este año. Queremos Calcular: (peso promedio de las zanahorias) Media, (peso central) Mediana, (Peso mas común)
Moda, (Qué tanto varían los pesos) Desviación estandar.

```python
import numpy as np
import statistics as stats
#Pesos de las zanahorias en gramos
zanahorias = [300, 280, 290, 310, 275, 290, 295, 315, 290, 280, 310, 305]
#Estadísticas descriptivas
media = np.mean(zanahorias) #promedio
mediana = np.median(zanahorias) #Valor Central
moda = stats.mode(zanahorias) #Valor mas común
desviacion = np.std(zanahorias) #Variabilidad
#Resultados :
print(f"Analisis de la cosecha de zanahorias ")
print(f"Media: {media}g")
print(f"Mediana: {mediana}g")
print(f"Moda: {moda}g")
print(f"Desviación estandar: {desviacion:.2f}g")
```

```python
import numpy as np
import statistics as stats
import matplotlib.pyplot as plt

# Función para analizar zanahorias
def analizar(zanahorias):
    # Estadísticas descriptivas
    media = np.mean(zanahorias)
    mediana = np.median(zanahorias)

    try:
        moda = stats.mode(zanahorias)
    except:
        moda = "No hay una única moda"

    desviacion = np.std(zanahorias)

    # Resultados
    print("\n🥕 Análisis de la cosecha de zanahorias 🥕")
    print(f"Media: {media:.2f} g")
    print(f"Mediana: {mediana:.2f} g")
    print(f"Moda: {moda}")
    print(f"Desviación estándar: {desviacion:.2f} g")

    # 📉 Visualización
    plt.figure()
    plt.hist(zanahorias, bins=6)
    plt.axvline(media)
    plt.title("Distribución de pesos de zanahorias")
    plt.xlabel("Peso (g)")
    plt.ylabel("Frecuencia")
    plt.show()


# Datos iniciales
zanahorias = [300, 280, 290, 310, 275, 290, 295, 315, 290, 280, 310, 305]

# Ejecutar análisis inicial
analizar(zanahorias)


# Entrada de nuevos datos
while True:
    opcion = input("\n¿Quieres ingresar nuevos pesos? (s/n): ").lower()

    if opcion == 's':
        entrada = input("Introduce los pesos separados por comas: ")
        try:
            nueva_lista = [float(x.strip()) for x in entrada.split(",")]
            if len(nueva_lista) < 2:
                print("Debes ingresar al menos dos valores.")
                continue
            analizar(nueva_lista)
        except ValueError:
            print("Error: introduce solo números válidos.")
    else:
        print("Fin del análisis de zanahorias.")
        break
```
