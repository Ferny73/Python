```python
#Sistema de busquedas por similitud
# Formato: [Acción, Comedia, SciFi, Drama, Terror]
cabecera_generos = ["Acción", "Comedia", "SciFi", "Drama", "Terror"]

lista_peliculas = {
    "Star Wars 1":    [1,0,1,0,0],
    "El Padrino":     [0,0,0,1,0],
    "Toy Story":      [0,1,0,0,0],
    "Casablanca":     [0,0,0,1,0],
    "Titanic":        [0,0,0,1,1],
    "Thriller":       [1,0,0,0,1],
    "Los Vengadores": [1,0,1,0,0],
    "Mario Galaxy":   [1,1,0,0,0],
    "Shrek":          [0,1,0,0,0],
    "Jurassic Park":  [0,0,1,0,1],
    "Indiana Jones":  [1,0,0,0,0],
    "Matrix":         [1,0,1,0,1],
    "Rambo":          [1,0,0,0,0]
}

def calcular_similitud(vector1, vector2):
    # Producto punto: suma de coincidencias
    return sum(a * b for a, b in zip(vector1, vector2))

def encontrar_palabras_clave(vectorP):
    # Retorna los nombres de los géneros que tienen un 1
    return [cabecera_generos[i] for i, valor in enumerate(vectorP) if valor == 1]

salida = False
#Bucle de interacción con el usuario
while not(salida):
  print("Buscador de Similitud en películas")
  print("Películas disponibles:", ", ".join(lista_peliculas.keys()))

  pelicula1 = input("\nElige la primera película: ").strip()
  pelicula2 = input("Elige la segunda película: ").strip()

  #Comparación de películas
  if pelicula1 in lista_peliculas and pelicula2 in lista_peliculas:
    vectorPeli1 = lista_peliculas[pelicula1]
    vectorPeli2 = lista_peliculas[pelicula2]

    puntos_similitud = calcular_similitud(vectorPeli1, vectorPeli2)
    etiquetaPeli1 = encontrar_palabras_clave(vectorPeli1)
    etiquetaPeli2 = encontrar_palabras_clave(vectorPeli2)

    print(f"\nResultados para: {pelicula1} vs {pelicula2}")
    print(f" Género de {pelicula1}: {etiquetaPeli1}")
    print(f" Género de {pelicula2}: {etiquetaPeli2}")
    print(f" Puntuación de Similitud: {puntos_similitud}")

    if puntos_similitud >1:
      print("¡Son muy parecidas!")
    elif puntos_similitud == 1:
      print("Tienen algún género en común")
    else:
      print("No se parecen en nada.")
  else:
    print("\n[!] Error: Una o ambas películas no están en la lista de películas.")

  #Continuar
  continuar = input("Deseas continuar (S/N)")
  if continuar == "S":
    salida = False
  elif continuar == "N":
    salida = True
  else:
    print("Entrada incorrecta, salimos!")
    salida = True
```
