```python
# =========================================
# VISUALIZACIÓN DE NOTAS DE ESTUDIANTES
# Matplotlib + Seaborn
# =========================================

# Librerías
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Estilo visual
sns.set_theme(style="whitegrid")

# =========================================
# CARGAR EL ARCHIVO CSV
# =========================================

# Subir archivo en Colab
from google.colab import files
uploaded = files.upload()

# Leer CSV
df = pd.read_csv("notas_estudiantes_completo (1).csv")

# Mostrar primeras filas
print("Primeros datos del dataset:")
print(df.head())

# =========================================
# TABLA ESTADÍSTICA
# =========================================

print("\nTabla estadística:")
print(df.describe())

# =========================================
# BARPLOT
# Nota media por asignatura
# =========================================

plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x="asignatura",
    y="nota",
    estimator="mean",
    palette="Set2"
)

plt.title("Nota media por asignatura")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# =========================================
# BOXPLOT
# Dispersión de notas
# =========================================

plt.figure(figsize=(10,5))

sns.boxplot(
    data=df,
    x="asignatura",
    y="nota",
    palette="pastel"
)

plt.title("Dispersión de notas por asignatura")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# =========================================
# COUNTPLOT
# Cantidad de registros
# =========================================

plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="asignatura",
    palette="viridis"
)

plt.title("Cantidad de registros por asignatura")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# =========================================
# HEATMAP
# Correlación entre variables numéricas
# =========================================

# Seleccionar columnas numéricas
variables = df.select_dtypes(include=["int64", "float64"])

# Correlación
corr = variables.corr()

plt.figure(figsize=(8,5))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlación entre variables")
plt.tight_layout()
plt.show()

# =========================================
# ANÁLISIS AUTOMÁTICO SIMPLE
# =========================================

print("\nANÁLISIS AUTOMÁTICO")

# Asignatura con mejor media
mejor_media = df.groupby("asignatura")["nota"].mean().idxmax()

# Asignatura con mayor variabilidad
mayor_variabilidad = df.groupby("asignatura")["nota"].std().idxmax()

print("Asignatura con mejor media:", mejor_media)
print("Asignatura con mayor variabilidad:", mayor_variabilidad)

# Mejores notas
print("\nTop 5 mejores notas:")
print(df.sort_values(by="nota", ascending=False).head())
```
